```python
STUDENT_ID="24/1035/BSCE-S"
COHORT="CML3112"
import csv
with open("site_inspection_log.csv",newline="") as f:
    rows=list(csv.DictReader(f))
```

## valid_count (5 marks)
Store an integer count of rows whose crack_width_mm field is nonblank. Skip blank readings before numeric conversion.


```python
valid_count = 0
for row in rows:
    if row['crack_width_mm'] != '':
        valid_count += 1
print(valid_count)
```

## Interesting Pattern
  This code counts how many crack width readings are not blank. It starts a counter at 0, goes through every row, and adds 1 whenever the reading is not empty. Blank readings are skipped, so the result is 297 out of 300 rows.

## mean_reading (5 marks)
Store the arithmetic mean of the nonblank crack_width_mm values as a number. Add the valid readings and divide by their count. Keep full precision.


```python
total_width = 0
for row in rows:
  if row['crack_width_mm'] != '':
    total_width += float(row['crack_width_mm'])
mean_width = total_width / valid_count
print(mean_width)
```

## Interesting Pattern
This code finds the average crack width. It adds up all the nonblank readings, converting each to a number with float, then divides the total by valid_count (297). Blanks are left out of both the total and the divisor. The result is 0.2262996632996631, kept at full precision.

## above_limit (5 marks)
Store an integer count of valid crack_width_mm readings strictly greater than 0.3. A reading equal to the threshold is not counted.


```python
above_limit = 0
for row in rows:
  if row['crack_width_mm'] != '' and float(row['crack_width_mm']) > 0.3:
    above_limit += 1
print(above_limit)
```

## Interesting Pattern
This code counts how many valid crack width readings are greater than 0.3 mm. It starts a counter at 0 and goes through every row. For each row, it first checks that the reading is not blank, then converts it to a number and checks if it is strictly greater than 0.3. If both conditions are true, it adds 1 to the counter. Blank readings are skipped, and a reading of exactly 0.3 is not counted. The final result is 59.

## group_means (5 marks)
Store a dictionary whose keys are the structure_id labels and whose values are the mean valid crack_width_mm readings within each group. Divide each group total by its own valid count. Keep full precision.


```python
group_totals = {}
group_counts = {}

# Accumulate totals and counts for each group, skipping blank values
for row in rows:
  structure_id = row['structure_id']
  crack_width_mm = row['crack_width_mm']
  if crack_width_mm != '':
    val = float(crack_width_mm)
    if structure_id not in group_totals:
      group_totals[structure_id] = 0.0
      group_counts[structure_id] = 0
      group_totals[structure_id] += val
      group_counts[structure_id] += 1

      # compute the mean for each group with full precision
      group_means ={}
      for struct_id in group_totals:
        group_means[struct_id] = round(group_totals[struct_id] / group_counts[struct_id], 2)

print(group_means)
```

## Interesting Pattern
This code finds the average crack width for each structure (B1, B2, B3...). It keeps two dictionaries: one with the running total of readings per structure, and one with how many valid readings each structure has. For every row, it skips blank readings, creates a new entry if the structure is new, then adds the reading to the total and 1 to the count. At the end, it divides each structure's total by its own count to get its mean.

Remove round(..., 2), since the question says to keep full precision.
